```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/content/Diwali Sales Data.csv', encoding='latin1')
df
```

```python
df.shape
```

```python
df.info()
```

```python
# threr are two null rows is there
df.drop(['Status','unnamed1'], axis=1, inplace= True) # remove them
df
```

```python
pd.isnull(df).sum()
```

```python
# in Amount there are 12 null valuse so we remove them
df.dropna(inplace=True)
```

```python
df.shape
```

```python
# for check the null
pd.isnull(df).sum()
```

```python
# change data type
# Ensure all values in 'Amount' are finite before converting to int
df = df[np.isfinite(df['Amount'])]
df['Amount'] = df['Amount'].astype('int')
```

```python
df.columns
```

```python
# rename the column
df.rename(columns={'Marital_Status': 'Shaadi'})
```

```python
# method returns description of the data in the data frame
df.describe()
```

```python
# use describe for specific column
df[['Age','Orders','Amount']].describe()
```

Exploratory Data Analysis

1. Gender


```python
df.columns
```

```python
gender = sns.countplot(x='Gender',data= df)
gender.bar_label(gender.containers[0])  # to see the exact number in the graph
```

```python
# how much you spent the money to buy
total_amt = df.groupby(['Gender'])['Amount'].sum().reset_index()
sns.barplot(x='Gender', y='Amount', data=total_amt)
```

```python
# Another easy approach: Using pandas' built-in plot method
# This aggregates the data and plots it directly in a single line.

# Calculate total amount spent by gender
gender_amount_sum = df.groupby('Gender')['Amount'].sum()

# Plot the results directly using pandas' plot method
gender_amount_sum.plot(kind='bar', figsize=(6, 4))
plt.title('Total Amount Spent by Gender (Pandas Plot)')
plt.xlabel('Gender')
plt.ylabel('Total Amount')


plt.show()
```

summary : Frome above graph we can see that most of the buyers are female instead of man.

2. AGE

```python
# which age group has more buyer power
ax = sns.countplot(data=df, x='Age Group', hue='Gender')
for bars in ax.containers:
    ax.bar_label(bars)
```

summary : there is 26-35 age group has more buying power

```python
# total amount vs age group
total_amt_age = df.groupby(['Age Group'])['Amount'].sum().reset_index()
sns.barplot(x='Age Group', y='Amount', data=total_amt_age)
```

: from above graph we can see that most buyers are of age group between 26-35 year female

STATE

```python
# total no of order from top 10 state
state = df.groupby(['State'])['Orders'].sum().reset_index()
state.head(10)
sns.set(rc={'figure.figsize':(15,5)})
sns.barplot(x='State', y='Orders', data=state.head(10))
```

: in above graph there is karnataka has more number of orders

```python
# total amount/sales from top 10 states
state = df.groupby(['State'])['Amount'].sum().reset_index()
state
sns.set(rc={'figure.figsize':(15,5)})
sns.barplot(x='State', y='Amount', data=state.head(10))
```

from above graph we can see that most of the order are from maharastra ,delhi, karnataka.

3.MARITIAL STATUS

```python
# maritial status
ax = sns.countplot(x='Marital_Status', data=df, color="grey")
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
# seprate by amount
marital = df.groupby(['Marital_Status','Gender'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
marital
sns.barplot(x='Marital_Status', y='Amount', data=marital, hue = 'Gender')
```

From above graph we can see that the married women has high purchesing power

4. OCCUPATION

```python
ax = sns.countplot(x='Occupation', data=df, palette='Set1')
```

in above graph we see that it sector, healthcare , banking, avation has most buying the things

```python
# Amount
occupation = df.groupby(['Occupation'])['Amount'].sum().reset_index()
sns.barplot(x='Occupation', y='Amount', data=occupation)
```

5. PRODUCT CATEGORY

```python
sns.set(rc={'figure.figsize':(20,5)})
ax = sns.countplot(x='Product_Category', data=df, palette='Set1')
ax
```

```python
# top 10 sold product sales
ax = df.groupby(['Product_ID'])['Orders'].sum().reset_index().sort_values(by='Orders', ascending=False).head(10)

sns.barplot(x='Product_ID', y='Orders', data=ax)
```

in above graph we can see that most sold products are food, footwear, and electronics.

CONCLUSION : Married women age group 26-35 yr from maharastra and karanataka, delhi working in IT , healthcare and avation are most likely to buy product from food clothing and electronic category
